```python
import torch

data = torch.load('E:/ITS/ISTS-PLM/data/activity/processed/data.pt')

# 递归查看数据结构
def explore_data(obj, indent=0):
    prefix = "  " * indent
    if isinstance(obj, dict):
        print(f"{prefix}字典，包含 {len(obj)} 个键:")
        for key, value in obj.items():
            print(f"{prefix}  '{key}':")
            explore_data(value, indent + 2)
    elif isinstance(obj, list):
        print(f"{prefix}列表，长度 {len(obj)}")
        if len(obj) > 0:
            print(f"{prefix}  第一个元素:")
            explore_data(obj[0], indent + 2)
    elif isinstance(obj, torch.Tensor):
        print(f"{prefix}Tensor: shape={obj.shape}, dtype={obj.dtype}")
    else:
        print(f"{prefix}{type(obj).__name__}: {obj}")

explore_data(data)
```

```python
import torch
import matplotlib.pyplot as plt
import numpy as np

# 读取数据
data = torch.load('E:/ITS/ISTS-PLM/data/activity/processed/data.pt', 
                  weights_only=False)

# 提取所有样本的原始数据和padding数据
original_lengths = []
padded_lengths = []
valid_ratios = []

for subject_id, timestamps, features, mask in data:
    # 原始数据长度（有效数据点数）
    valid_mask = mask.sum(dim=1) > 0  # 每行至少有一个有效值
    original_length = valid_mask.sum().item()
    original_lengths.append(original_length)
    
    # Padding后的长度
    padded_length = features.shape[0]
    padded_lengths.append(padded_length)
    
    # 有效数据比例
    valid_ratio = original_length / padded_length
    valid_ratios.append(valid_ratio)

# 创建图表
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('数据集分布分析：原始 vs Padding', fontsize=16, fontweight='bold')

# 1. 序列长度分布对比（直方图）
ax1 = axes[0, 0]
bins = np.linspace(0, max(padded_lengths), 30)
ax1.hist(original_lengths, bins=bins, alpha=0.6, label='原始数据', color='blue', edgecolor='black')
ax1.hist(padded_lengths, bins=bins, alpha=0.6, label='Padding后', color='red', edgecolor='black')
ax1.set_xlabel('序列长度', fontsize=12)
ax1.set_ylabel('样本数量', fontsize=12)
ax1.set_title('序列长度分布对比', fontsize=13, fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# 2. 密度分布图
ax2 = axes[0, 1]
from scipy import stats
original_density = stats.gaussian_kde(original_lengths)
padded_density = stats.gaussian_kde(padded_lengths)
x_range = np.linspace(0, max(padded_lengths), 500)
ax2.plot(x_range, original_density(x_range), label='原始数据', color='blue', linewidth=2)
ax2.plot(x_range, padded_density(x_range), label='Padding后', color='red', linewidth=2)
ax2.fill_between(x_range, original_density(x_range), alpha=0.3, color='blue')
ax2.fill_between(x_range, padded_density(x_range), alpha=0.3, color='red')
ax2.set_xlabel('序列长度', fontsize=12)
ax2.set_ylabel('密度', fontsize=12)
ax2.set_title('密度分布对比', fontsize=13, fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3)

# 3. 有效数据比例分布
ax3 = axes[1, 0]
ax3.hist(valid_ratios, bins=20, color='green', alpha=0.7, edgecolor='black')
ax3.set_xlabel('有效数据比例', fontsize=12)
ax3.set_ylabel('样本数量', fontsize=12)
ax3.set_title('有效数据占比分布', fontsize=13, fontweight='bold')
ax3.axvline(np.mean(valid_ratios), color='red', linestyle='--', 
            linewidth=2, label=f'平均值: {np.mean(valid_ratios):.2%}')
ax3.legend()
ax3.grid(True, alpha=0.3)

# 4. 每个样本的长度对比（条形图）
ax4 = axes[1, 1]
sample_indices = range(len(data))
width = 0.35
ax4.bar([i - width/2 for i in sample_indices], original_lengths, 
        width, label='原始数据', color='blue', alpha=0.7)
ax4.bar([i + width/2 for i in sample_indices], padded_lengths, 
        width, label='Padding后', color='red', alpha=0.7)
ax4.set_xlabel('样本索引', fontsize=12)
ax4.set_ylabel('序列长度', fontsize=12)
ax4.set_title('每个样本的长度对比', fontsize=13, fontweight='bold')
ax4.legend()
ax4.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.savefig('data_distribution_comparison.png', dpi=300, bbox_inches='tight')
plt.show()

# 打印统计信息
print("="*60)
print("数据统计信息")
print("="*60)
print(f"总样本数: {len(data)}")
print(f"\n原始数据:")
print(f"  平均长度: {np.mean(original_lengths):.2f}")
print(f"  最小长度: {np.min(original_lengths)}")
print(f"  最大长度: {np.max(original_lengths)}")
print(f"  标准差: {np.std(original_lengths):.2f}")
print(f"\nPadding后:")
print(f"  统一长度: {padded_lengths[0]}")
print(f"\n有效数据比例:")
print(f"  平均: {np.mean(valid_ratios):.2%}")
print(f"  最小: {np.min(valid_ratios):.2%}")
print(f"  最大: {np.max(valid_ratios):.2%}")
print(f"\nPadding增加:")
print(f"  平均增加: {np.mean(padded_lengths) - np.mean(original_lengths):.2f} 个时间步")
print(f"  平均padding比例: {(1 - np.mean(valid_ratios)):.2%}")
print("="*60)
```

```python
import torch
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import rcParams

# 设置中文字体
rcParams['font.sans-serif'] = ['SimHei']  # 用黑体
rcParams['axes.unicode_minus'] = False  # 解决负号显示问题

# 读取数据
data = torch.load('E:/ITS/ISTS-PLM/data/activity/processed/data.pt', 
                  weights_only=False)

# 统计信息
stats = {
    'subject_ids': [],
    'total_timesteps': [],
    'total_features': [],
    'valid_features': [],
    'valid_feature_ratio': [],
    'avg_valid_features_per_timestep': [],
    'min_valid_features_per_timestep': [],
    'max_valid_features_per_timestep': []
}

for subject_id, timestamps, features, mask in data:
    total_timesteps = features.shape[0]
    total_features = features.shape[1]
    valid_features = (mask == 1).sum().item()
    
    valid_per_timestep = mask.sum(dim=1)
    
    stats['subject_ids'].append(subject_id)
    stats['total_timesteps'].append(total_timesteps)
    stats['total_features'].append(total_features)
    stats['valid_features'].append(valid_features)
    stats['valid_feature_ratio'].append(valid_features / (total_timesteps * total_features))
    stats['avg_valid_features_per_timestep'].append(valid_per_timestep.mean().item())
    stats['min_valid_features_per_timestep'].append(valid_per_timestep.min().item())
    stats['max_valid_features_per_timestep'].append(valid_per_timestep.max().item())

# 创建图表
fig = plt.figure(figsize=(16, 12))
gs = fig.add_gridspec(3, 3, hspace=0.3, wspace=0.3)

# 总标题
fig.suptitle('数据集特征维度 Padding 分析', fontsize=18, fontweight='bold')

# 1. 时间步数分布
ax1 = fig.add_subplot(gs[0, 0])
ax1.bar(range(len(data)), stats['total_timesteps'], color='steelblue', alpha=0.7)
ax1.set_xlabel('样本索引', fontsize=11)
ax1.set_ylabel('时间步数', fontsize=11)
ax1.set_title('各样本的时间步数', fontsize=12, fontweight='bold')
ax1.grid(True, alpha=0.3, axis='y')
ax1.text(0.95, 0.95, f"平均: {np.mean(stats['total_timesteps']):.0f}\n"
         f"范围: {np.min(stats['total_timesteps'])}-{np.max(stats['total_timesteps'])}",
         transform=ax1.transAxes, ha='right', va='top',
         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# 2. 有效特征占比
ax2 = fig.add_subplot(gs[0, 1])
colors = plt.cm.RdYlGn(stats['valid_feature_ratio'])
ax2.bar(range(len(data)), [r*100 for r in stats['valid_feature_ratio']], 
        color=colors, alpha=0.8)
ax2.axhline(np.mean(stats['valid_feature_ratio'])*100, color='red', 
            linestyle='--', linewidth=2, label=f'平均: {np.mean(stats["valid_feature_ratio"])*100:.1f}%')
ax2.set_xlabel('样本索引', fontsize=11)
ax2.set_ylabel('有效特征占比 (%)', fontsize=11)
ax2.set_title('有效特征 vs Padding 比例', fontsize=12, fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3, axis='y')

# 3. 每时间步平均有效特征数
ax3 = fig.add_subplot(gs[0, 2])
ax3.bar(range(len(data)), stats['avg_valid_features_per_timestep'], 
        color='coral', alpha=0.7)
ax3.set_xlabel('样本索引', fontsize=11)
ax3.set_ylabel('平均有效特征数', fontsize=11)
ax3.set_title(f'每时间步的有效特征数 (总共{stats["total_features"][0]}个)', 
              fontsize=12, fontweight='bold')
ax3.grid(True, alpha=0.3, axis='y')
ax3.text(0.95, 0.95, f"平均: {np.mean(stats['avg_valid_features_per_timestep']):.2f}",
         transform=ax3.transAxes, ha='right', va='top',
         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# 4. 有效特征数分布（饼图）
ax4 = fig.add_subplot(gs[1, 0])
total_cells = sum([t * f for t, f in zip(stats['total_timesteps'], stats['total_features'])])
total_valid = sum(stats['valid_features'])
total_padding = total_cells - total_valid
sizes = [total_valid, total_padding]
labels = [f'有效数据\n{total_valid:,}\n({total_valid/total_cells*100:.1f}%)', 
          f'Padding\n{total_padding:,}\n({total_padding/total_cells*100:.1f}%)']
colors_pie = ['#66c2a5', '#fc8d62']
explode = (0.05, 0)
ax4.pie(sizes, explode=explode, labels=labels, colors=colors_pie,
        autopct='', shadow=True, startangle=90)
ax4.set_title('整体数据：有效 vs Padding', fontsize=12, fontweight='bold')

# 5. 每时间步有效特征数的范围
ax5 = fig.add_subplot(gs[1, 1])
x = range(len(data))
ax5.bar(x, stats['max_valid_features_per_timestep'], 
        label='最大值', color='lightgreen', alpha=0.6)
ax5.bar(x, stats['avg_valid_features_per_timestep'], 
        label='平均值', color='green', alpha=0.8)
ax5.bar(x, stats['min_valid_features_per_timestep'], 
        label='最小值', color='darkgreen', alpha=0.9)
ax5.set_xlabel('样本索引', fontsize=11)
ax5.set_ylabel('有效特征数', fontsize=11)
ax5.set_title('每样本中时间步的有效特征数范围', fontsize=12, fontweight='bold')
ax5.legend()
ax5.grid(True, alpha=0.3, axis='y')

# 6. 有效特征占比分布直方图
ax6 = fig.add_subplot(gs[1, 2])
ax6.hist([r*100 for r in stats['valid_feature_ratio']], bins=15, 
         color='purple', alpha=0.7, edgecolor='black')
ax6.axvline(np.mean(stats['valid_feature_ratio'])*100, color='red', 
            linestyle='--', linewidth=2, label=f'平均值')
ax6.set_xlabel('有效特征占比 (%)', fontsize=11)
ax6.set_ylabel('样本数', fontsize=11)
ax6.set_title('有效特征占比分布', fontsize=12, fontweight='bold')
ax6.legend()
ax6.grid(True, alpha=0.3)

# 7. 单个样本示例可视化
ax7 = fig.add_subplot(gs[2, :])
sample_idx = 0  # 第一个样本
subject_id, timestamps, features, mask = data[sample_idx]
# 只显示前200个时间步
display_steps = min(200, features.shape[0])
mask_sample = mask[:display_steps].cpu().numpy()

im = ax7.imshow(mask_sample.T, aspect='auto', cmap='RdYlGn', 
                interpolation='nearest', vmin=0, vmax=1)
ax7.set_xlabel('时间步', fontsize=11)
ax7.set_ylabel('特征维度', fontsize=11)
ax7.set_title(f'样本 {subject_id} 的前{display_steps}个时间步的Mask可视化 '
              f'(绿色=有效数据, 红色=Padding)', fontsize=12, fontweight='bold')
ax7.set_yticks(range(features.shape[1]))
ax7.set_yticklabels([f'F{i}' for i in range(features.shape[1])])

# 添加颜色条
cbar = plt.colorbar(im, ax=ax7, orientation='vertical', pad=0.01)
cbar.set_label('数据有效性', fontsize=10)
cbar.set_ticks([0, 1])
cbar.set_ticklabels(['Padding', '有效'])

plt.savefig('feature_padding_analysis.png', dpi=300, bbox_inches='tight')
plt.show()

# 打印详细统计
print("\n" + "="*70)
print("数据集详细统计")
print("="*70)
print(f"总样本数: {len(data)}")
print(f"特征维度: {stats['total_features'][0]}")
print(f"\n时间步统计:")
print(f"  平均时间步数: {np.mean(stats['total_timesteps']):.2f}")
print(f"  时间步范围: {np.min(stats['total_timesteps'])} - {np.max(stats['total_timesteps'])}")
print(f"\n有效特征统计:")
print(f"  平均每时间步有效特征数: {np.mean(stats['avg_valid_features_per_timestep']):.2f} / {stats['total_features'][0]}")
print(f"  有效特征占比: {np.mean(stats['valid_feature_ratio'])*100:.2f}%")
print(f"  Padding占比: {(1-np.mean(stats['valid_feature_ratio']))*100:.2f}%")
print(f"\n总数据量:")
print(f"  总单元格数: {total_cells:,}")
print(f"  有效数据: {total_valid:,} ({total_valid/total_cells*100:.2f}%)")
print(f"  Padding数据: {total_padding:,} ({total_padding/total_cells*100:.2f}%)")
print("="*70)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

def visualize_ot_plan(transport_plan, z_source, z_target):
    """
    可视化流形上的传输计划
    """
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    
    # 1. 传输矩阵热图
    ax = axes[0]
    sns.heatmap(
        transport_plan.cpu().numpy(),
        cmap='Blues',
        ax=ax,
        cbar_kws={'label': 'Transport Mass'}
    )
    ax.set_title('Optimal Transport Plan')
    ax.set_xlabel('Target Samples')
    ax.set_ylabel('Source Samples')
    
    # 2. 流形空间可视化（用PCA降到2D）
    from sklearn.decomposition import PCA
    
    z_source_np = z_source.detach().cpu().numpy()
    z_target_np = z_target.detach().cpu().numpy()
    
    pca = PCA(n_components=2)
    z_all = np.vstack([z_source_np, z_target_np])
    z_2d = pca.fit_transform(z_all)
    
    z_source_2d = z_2d[:len(z_source_np)]
    z_target_2d = z_2d[len(z_source_np):]
    
    ax = axes[1]
    ax.scatter(z_source_2d[:, 0], z_source_2d[:, 1], 
               c='red', alpha=0.6, label='Incomplete (Source)', s=50)
    ax.scatter(z_target_2d[:, 0], z_target_2d[:, 1], 
               c='blue', alpha=0.6, label='Complete (Target)', s=50)
    
    # 画传输箭头（只画前10个）
    T_np = transport_plan.cpu().numpy()
    for i in range(min(10, len(z_source_2d))):
        # 找主要传输目标
        top_targets = np.argsort(T_np[i])[-3:]
        for j in top_targets:
            if T_np[i, j] > 0.01:  # 只画显著的传输
                ax.arrow(
                    z_source_2d[i, 0], z_source_2d[i, 1],
                    z_target_2d[j, 0] - z_source_2d[i, 0],
                    z_target_2d[j, 1] - z_source_2d[i, 1],
                    alpha=T_np[i, j] * 5,
                    head_width=0.1,
                    color='green'
                )
    
    ax.set_title('Manifold Space & Transport')
    ax.legend()
    
    # 3. Wasserstein距离收敛曲线
    ax = axes[2]
    # (需要在训练时记录)
    ax.set_title('Wasserstein Distance over Training')
    ax.set_xlabel('Epoch')
    ax.set_ylabel('W2 Distance')
    
    plt.tight_layout()
    plt.savefig('ot_visualization.png', dpi=150)
    plt.show()
```
